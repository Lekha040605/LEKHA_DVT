```python
import pandas as pd
df = pd.read_csv("student_data.csv")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_marks = df[df["Internal_Marks"] > 85]
print(high_marks)
```

```python
it_students = df[df["Department"]=="IT"]
print(it_students)
```

```python
sorted_df = df.sort_values(
    by="Internal_Marks",
    ascending=False
)
print(sorted_df)
```

```python
print("Mean =",
      df["Internal_Marks"].mean())

print("Median =",
      df["Internal_Marks"].median())

print("Mode =",
      df["Internal_Marks"].mode())

print("Standard Deviation =",
      df["Internal_Marks"].std())
```

```python
print(df.describe())
```

```python
corr = df[[
"External_Marks",
"Internal_Marks"
]].corr()

print(corr)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Name"], df["Internal_Marks"])

plt.title("Internal Marks of Students")
plt.xlabel("Student Name")
plt.ylabel("Internal Marks")

plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(df["External_Marks"])

plt.title("Box Plot of External Marks")
plt.ylabel("Marks")

plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df["Internal_Marks"], df["External_Marks"])

plt.title("Internal vs External Marks")
plt.xlabel("Internal Marks")
plt.ylabel("External Marks")

plt.show()
```
