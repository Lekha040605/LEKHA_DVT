```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales = [120,150,180,170,220,250]

plt.plot(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales = [120,150,180,170,220,250]

plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

departments = ["HR","IT","Finance","Sales"]
employees = [25,40,20,15]

plt.pie(
    employees,
    labels=departments,
    autopct='%1.1f%%'
)

plt.title("Employee Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45,50,55,60,65,65,70,70,75,80,85,90,95]

plt.hist(marks, bins=5)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

attendance = [60,65,70,75,80,85,90,95]
marks = [50,55,60,65,70,75,80,90]

plt.scatter(attendance, marks)
plt.title("Attendance vs Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

salary = [25000,27000,30000,32000,35000,40000,150000]

plt.boxplot(salary)
plt.title("Salary Distribution")
plt.ylabel("Salary")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr"]
sales = [120,150,180,200]
profit = [20,30,40,50]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.title("Sales and Profit")
plt.xlabel("Months")
plt.ylabel("Value")
plt.legend()
plt.show()
```
